```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = sns.load_dataset("titanic")

# Basic data cleaning
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df = df.drop("deck", axis=1)
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="survived")

plt.title("Titanic Passenger Survival Distribution")
plt.xlabel("Survival Status (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
gender_survival = df.groupby("sex")["survived"].mean() * 100

plt.figure(figsize=(7, 5))
sns.barplot(x=gender_survival.index, y=gender_survival.values)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")

plt.show()
```

```python
class_survival = df.groupby("pclass")["survived"].mean() * 100

plt.figure(figsize=(7, 5))
sns.barplot(x=class_survival.index, y=class_survival.values)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["age"], bins=30, kde=True)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="pclass", y="fare")

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

numeric_df = df.select_dtypes(include=["number"])

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Titanic Dataset")

plt.show()
```

```python
alone_survival = df.groupby("alone")["survived"].mean() * 100

plt.figure(figsize=(7, 5))
sns.barplot(x=alone_survival.index, y=alone_survival.values)

plt.title("Survival Rate: Alone vs With Family")
plt.xlabel("Travelling Alone")
plt.ylabel("Survival Rate (%)")

plt.show()
```
